```python
num = input("enter the values:").split()
print(num)
num1=[]
for x in num:
    num1.append(int(x))
print(num1)
maxi=0
mini=0
sumi=0

for x in num1:
    sumi=sumi+x
    if maxi<x:
        maxi=x
    if mini>x:
        mini=x
    
print(maxi)
print(mini)
print(sumi)
    
```
