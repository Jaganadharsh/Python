```python
#Float Division
x=21
y=4
div= x/y
print('x= ',x)
print('y= ',y)
print('Float Div=',div)
```

```python
#Floor Division
import math
x=21
y=4
div= x/y
print('x= ',x)
print('y= ',y)
print('Float Div=',math.floor(div))
```

```python
#Modulo
x=21
y=4
div= x%y
print('x= ',x)
print('y= ',y)
print('Modulo=', div)
```

```python
#Power
import math
x=2
y=4
print('x= ',x)
print('y= ',y)
print('Power=',math.pow(x,y))
print('Power=',x ** y)
```

```python
#Power with out functions(using for loop)
import math
x=2
y=4
z=1
for i in range (1,y+1,1):
    z=z*x
    
print('x= ',x)
print('y= ',y)
print('Power=',z)
```
