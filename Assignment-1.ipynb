```python
weight=int(input('Enter the BMI Index:'))
if weight < 18.5:
    print('Under weight')
elif weight < 24.9:
    print('Normal')
elif weight < 29.9:
    print('Overweight')
elif weight < 34.9:
    print('Obesity (Class I) or Very Overweight')
elif weight < 39.9:
    print('Obesity (Class II)')
else:
    print('Extreme obesity')
        
```
