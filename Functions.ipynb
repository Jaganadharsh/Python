```python
# factorial recursive

def factc(n):
    
    if n!=0:
        global cancat
        cancat = cancat+ str(n)+"*"
        return n * factc(n-1)
      
    else:
        return 1
    

print(cancat,factc(5)) # cancat wont print , since factc function execute firest with initial cancat value
print(cancat)


cancat=''
    
```

```python
def factc(n,cancat=''):
    
    if n!=0:
        cancat+=str(n)+"*"
        f,c=factc(n-1,cancat)
        return n * f, c
      
    else:
        return 1,cancat
    

print(factc(5))
```

```python
#comprehension
x=[1,2,3,4,5]
s=[]
for i in x:
    s.append(i ** 2)
print(s)
x.append(6)
print(s)
s=[i**2 for i in x]
print(s)
x.append(7)
print(s)
```

```python
#comprehension condition
x=[1,2,3,4,5]
s=[i for i in x if i%2==0]
print(s)
x=[1,2,3,4,5]
s=[i if i%2==0 else "odd" for i in x]
print(s)
```

```python
# combined comprehension matrix
m =[[1,2],[2,3]]
print(m)
n=[i for row in m for i in row]

print(n)
    
```
